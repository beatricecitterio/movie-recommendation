```python
import tkinter 
import customtkinter

# System Settings
customtkinter.set_appearance_mode("System" ) 
customtkinter.set_default_color_theme("blue")

# Our app frame
app = customtkinter.CTk()
app.geometry("720x480" )
app.title("Movie Reco")

# Adding UI Elements
title = customtkinter.CTkLabel(app, text="Which genre are you in the mood for?")
title.pack(padx=10, pady=10)
# Link input
input_genre = tkinter.StringVar()
genre = customtkinter.CTkEntry(app, width=350, height=40, textvariable=input_genre)
genre.pack()

# Run app
app.mainloop()
```
