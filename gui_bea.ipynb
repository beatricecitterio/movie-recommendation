```python
import tkinter as tk
from tkinter import ttk, messagebox
import pandas as pd
import numpy as np
from suggestion import movies_to_rate, suggestion

# Move to rating page
def open_rating_page():
    genre = genre_var.get()
    num_ratings = num_ratings_var.get()

    # Clear previous entries
    for widget in root.winfo_children():
        widget.destroy()

    # Movie Ratings Input
    global rating_entries
    rating_entries = {}
    tk.Label(root, text="Rate the following movies (0-5, 0.5 increments, or 'Not seen'):").pack(pady=5)
    movies = movies_to_rate(genre, num_ratings)
    for movie in movies:
        tk.Label(root, text=movie).pack(pady=2)
        entry = tk.Entry(root)
        entry.pack(pady=2)
        rating_entries[movie] = entry

    # Number of Recommendations Selection
    tk.Label(root, text="Choose number of recommendations:").pack(pady=5)
    global num_suggestions_var
    num_suggestions_var = tk.IntVar()
    num_suggestions_var.set(3)
    tk.Radiobutton(root, text="1", variable=num_suggestions_var, value=1).pack()
    tk.Radiobutton(root, text="3", variable=num_suggestions_var, value=3).pack()
    tk.Radiobutton(root, text="5", variable=num_suggestions_var, value=5).pack()

    # Recommend Button
    tk.Button(root, text="Next", command=process_ratings).pack(pady=20)

# Process ratings before moving to recommendation page
def process_ratings():
    global user_ratings
    user_ratings = []
    try:
        for entry in rating_entries.values():
            rating = entry.get()
            if rating == 'Not seen':
                user_ratings.append('Not seen')
            else:
                float_rating = float(rating)
                if 0 <= float_rating <= 5 and float_rating % 0.5 == 0:
                    user_ratings.append(float_rating)
                else:
                    raise ValueError
    except ValueError:
        messagebox.showerror("Input Error", "Ratings must be between 0 and 5, in 0.5 increments, or 'Not seen'.")
        return
    open_recommendation_page()

# Move to recommendation page
def open_recommendation_page():
    # Clear previous entries
    for widget in root.winfo_children():
        widget.destroy()

    # Get recommendations using the provided suggestion function
    genre = genre_var.get()
    num_suggestions = num_suggestions_var.get()
    recommended_movies = suggestion(user_ratings, genre, len(user_ratings), num_suggestions)

    # Display recommendations
    tk.Label(root, text="Recommended Movies:").pack(pady=5)
    for movie in recommended_movies:
        tk.Label(root, text=movie).pack(pady=2)



# GUI Window Setup
root = tk.Tk()
root.title("Movie Recommendation System")
root.geometry("400x400")

# First Page - Genre and Number of Ratings Selection
tk.Label(root, text="Select Genre:").pack(pady=5)
genre_var = tk.StringVar()
genres = ['Action', 'Adventure', 'Animation', 'Children', 'Comedy', 'Crime', 'Documentary', 'Drama', 'Fantasy', 'Film-Noir', 'Horror', 'Musical', 'Mystery', 'Romance', 'Sci-Fi', 'Thriller', 'War', 'Western', 'Any Genre']
genre_dropdown = ttk.Combobox(root, textvariable=genre_var, values=genres, state='readonly')
genre_dropdown.pack(pady=5)
genre_dropdown.current(0)

# Number of Ratings Selection
tk.Label(root, text="Choose number of movies to rate:").pack(pady=5)
num_ratings_var = tk.IntVar()
num_ratings_var.set(5)
tk.Radiobutton(root, text="5", variable=num_ratings_var, value=5).pack()
tk.Radiobutton(root, text="10", variable=num_ratings_var, value=10).pack()

# Next Button
tk.Button(root, text="Next", command=open_rating_page).pack(pady=20)

# Start GUI Main Loop
root.mainloop()
```
